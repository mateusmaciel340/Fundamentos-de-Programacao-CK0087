## 2. Regras para nomear variáveis

No Python, não podemos dar qualquer nome para as nossas variáveis. Existem algumas regras que a linguagem exige.

### Regras obrigatórias:
O nome da variável
1.  Não pode começar com número.
2.  Não pode conter espaços em branco ou símbolos especiais (como `@`, `$`, `%`, `!`, hífens `-`, emojis, e muitos outros).
3.  Não pode ser palavra reservadas pela linguagem Python (como `print`, `if`, `True`).


### Boas práticas:
*   Utilize nomes descritivos (use `preco_produto` em vez de apenas `p`).
*   Para nomes compostos, use letras minúsculas separadas por sublinhado (estilo chamado `snake_case`), como `nota_final`, `nome_completo`, `limite_cartao`.

> **Sobre acentos e cedilha (`ç`):** O Python permite o uso de caracteres acentuados (como `média = 7.5`) ou com cedilha (como `preço = 11`). No entanto, na prática, evita-se usar acentos ou cedilhas em nomes de variáveis. Isso previne problemas ou inconveniências quando o código é compartilhado com pessoas de outros países (que podem não ter o seu teclado) ou simplesmente sistemas que não esperam esses símbolos especiais.

## 3. Tipos primitivos de dados

Cada dado que guardamos em uma variável pertence a um tipo. O Python identifica esse tipo automaticamente. Os quatro tipos básicos mais importantes são:

*   **`int`** (Inteiros): Números sem parte decimal (ex: `15`, `-3`, `0`).
*   **`float`** (Ponto flutuante): Números com parte decimal, utilizando ponto em vez de vírgula (ex: `5.99`, `-1.5`).
*   **`str`** (String/texto): Textos delimitados por aspas simples ou duplas (ex: `"Ana"`, `'Python'`).
*   **`bool`** (Booleano): Valores lógicos, que só podem ser `True` (Verdadeiro) ou `False` (Falso).

Podemos usar a instrução `type()` para perguntar ao Python qual é o tipo de uma variável.

As aspas especificam que um valor deve ser interpretado como texto. `10` é um número inteiro, mas `"10"` é um texto; `True` é um valor lógico, mas `"True"` é um texto. O `type()` mostra essa diferença.

Talvez você repare que `print(type(...))` exibe algo como `<class 'int'>`. A palavra `class` aqui é apenas a forma que o Python usa para se referir aos seus tipos internos; não precisa se preocupar com ela neste momento.

````python
# Criando variáveis de diferentes tipos
quantidade_itens = 10
valor_unitario = 4.50
nome_produto = "Caderno"
estoque_disponivel = True

# Verificando o tipo de cada uma
print(type(quantidade_itens)) # type(10) => int
print(type(valor_unitario)) # type(4.50) => float 
print(type(nome_produto)) # type("Caderno") => str
print(type(estoque_disponivel)) # type(True) => bool
````

````python
# Repare na importância das aspas:
print(type(10)) # => int
print(type("10")) # str
print(type(True)) # bool
print(type("True")) # str
````

Quando o computador encontra `uma_variavel = alguma_coisa`, ele primeiro calcula o valor do lado direito do `=` e só depois guarda esse resultado na variável da esquerda. Por isso podemos usar a própria variável no cálculo.


###
Imagine que estamos contando questionários de uma pesquisa que vão chegando um a um. A cada questionário recebido, somamos 1 à contagem anterior:

````python
questionarios = 0
print("Questionários até agora:", questionarios)

# incremente questionarios aqui
questionarios = questionarios + 1
print("Chegou mais um! Questionários até agora:", questionarios)

# incremente questionarios aqui
questionarios = questionarios + 1
print("Chegou mais um! Questionários até agora:", questionarios)

# incremente questionarios aqui
questionarios = questionarios + 1
print("Chegou mais um! Questionários até agora:", questionarios)
````

###
Em `questionarios = questionarios + 1`, o Python primeiro olha o lado direito. Se `questionarios` vale 0, calcula `0 + 1` e guarda esse resultado (1) na variável. Na vez seguinte, o valor já é 1, então calcula `1 + 1` e guarda 2, e assim por diante. Essa ideia é estranha para quem vem da matemática (lá, `x = x + 1` não faz sentido), mas é comum em programação.

O ponto principal é que o símbolo `=` em Python representa *atribuição* e não igualdade. Atribuição é uma operação que não é simétrica: x = 1 faz sentido em Python, mas 1 = x não, pois o lado esquerdo de uma atribuição deve ser a variável que receberá o valor da expressão à direita.

## 5. Operadores aritméticos

Para realizar cálculos matemáticos com variáveis numéricas, utilizamos os seguintes operadores:

| Operador | Operação | Exemplo |
| :---: | :--- | :--- |
| `+` | Adição | `a + b` |
| `-` | Subtração | `a - b` |
| `*` | Multiplicação | `a * b` |
| `/` | Divisão | `a / b` |
| `//` | Divisão Inteira (trunca as casas decimais) | `a // b` |
| `%` | Resto da Divisão (módulo) | `a % b` |
| `+` (unário) | Positivo | `+a` |
| `-` (unário) | Negativo | `-a` |
| `**` | Exponenciação (potência) | `a ** b` |


````python
a = 10
b = 3

print("Soma:", a + b) # 10 + 3 = 13
print("Subtração:", a - b) # 10 - 3 = 7
print("Multiplicação:", a * b) # 10 * 3 = 30
print("Divisão:", a / b) # 10 / 3 = 3.333..
print("Divisão inteira:", a // b) # 10 // 3 = 3
print("Resto da divisão:", a % b) # 10 % 3 => 10 = 3 * 3 + r => r = 10 - 9 => r = 1
print("Exponenciação:", a ** b) # 10 ** 3 = 1000
````

###
Divisão inteira `//` e resto `%` são bem úteis em situações do dia a dia. Suponha que 83 pessoas precisem ser divididas em grupos de 5. Quantos grupos completos conseguimos formar? Sobram quantas pessoas? A divisão inteira responde a primeira pergunta e o resto responde a segunda:

````python
pessoas = 83
tamanho_grupo = 5

grupos_completos = pessoas // tamanho_grupo
pessoas_sobrando = pessoas % tamanho_grupo

# 83 // 5 = 16 * 5 => 32 + 32 = 64 + 16 = 60 + 10 + 10 = 80, grupos_completos = 16
print("Grupos completos:", grupos_completos)

# 83 = 5 * 16 + r => r = 83 - 80 => r = 3
print("Pessoas sobrando:", pessoas_sobrando)
````

## 6. Precedência de operadores

Quando usamos várias operações em uma única linha, o Python resolve a expressão seguindo uma ordem de prioridade padrão inspirada pelas regras matemáticas usuais. Do mais prioritário para o menos prioritário:

1. Parênteses `()`.
2. Exponenciação `**`.
3. Indicador de sinal (`+` e `-` unários).
4. Multiplicação `*`, divisões `/`, `//` e resto `%`.
5. Adição `+` e subtração `-`.

E quando dois operadores têm a mesma prioridade, como em `10 / 2 * 3`? O Python resolve da esquerda para a direita, como fazemos na matemática. Então `10 / 2 * 3` é `(10 / 2) * 3`, e não `10 / (2 * 3)`. A exceção é o operador de exponenciação `**`, que resolve da direita para a esquerda (dizemos que tem associatividade à direita).

````python
# Caso 1: Sem parênteses. A multiplicação (3 * 2) é feita antes da soma.
resultado_calculo_um = 5 + 3 * 2 # 5 + 6 => resultado_calculo-um = 11
print("Resultado do caso 1:", resultado_calculo_um)

# Caso 2: Com parênteses. A soma (5 + 3) é feita antes da multiplicação.
resultado_calculo_dois = (5 + 3) * 2 # (8) * 2 => resultado_calculo_dois = 16
print("Resultado do caso 2:", resultado_calculo_dois)
````

###
Um caso muito comum em que os parênteses mudam tudo é a média de duas notas. A fórmula correta é `(nota1 + nota2) / 2`. Sem os parênteses, teríamos `nota1 + (nota2 / 2)`, um resultado completamente diferente:

````python
nota_prova_um = 7.5
nota_prova_dois = 8.5
media_certa = (nota_prova_um + nota_prova_dois) / 2

# Com parênteses: primeiro soma, depois divide por 2
# media_certa = ??
print("Com parênteses:", media_certa)

# Sem parênteses: a divisão por 2 acontece antes da soma
media_errada = nota_prova_um + nota_prova_dois / 2
print("Sem parênteses:", media_errada)
````

## Bônus: atribuição composta
Já aprendemos a fazer atribuição (`x = 2`) e também operações aritméticas (`x + 2`).

###
Um caso comum é fazer uma atribuição que usa o próprio alvo da atribuição na operação aritmética:

````python
x = x + 2
````

###

Nesse caso, há um atalho conveniente:

````python
x = 4

x += 2 # mesma coisa que x = x + 2
print(f"x = {4} + {2} = {x}") # x = 4 + 2 => x = 6
````

###
Similarmente,

````python
x = 12

x += 2  # mesma coisa que x = x + 2
print(f"x = {12} + {2} => {x}") # x = 12 + 2 => x = 14

x -= 2  # mesma coisa que x = x - 2
print(f"x = {14} - {2} => {x}") # x = 14 - 2 => x = 12

x *= 2  # mesma coisa que x = x * 2
print(f"x = {12} * {2} => {x}") # x = 12 * 2 => x = 24

x /= 2  # mesma coisa que x = x / 2
print(f"x = {24} / {2} => {x}") # 24 / 2 => x = 12.0

x //= 2 # mesma coisa que x = x // 2
print(f"x = {12.0} // {2} => {x}") # 12.0 // 2 => 6.0

x %= 2  # mesma coisa que x = x % 2
print(f"x = {6.0} % {2} => {x}") # 6.0 = 2 * 3 + r => r = 6.0 - 6 => r = 0.0, x = 0.0
````

### Exercício 11: Precedência e associatividade

Vimos que `**` tem prioridade sobre `*`, `/`, `//`, `%`, que por sua vez têm prioridade sobre `+` e `-`. Já a exponenciação é associativa à direita, ou seja, `2 ** 3 ** 2` é interpretado como `2 ** (3 ** 2)`, e não `(2 ** 3) ** 2`. Os outros operadores (`+`, `-`, `*`, `/`, `//`, `%`) são associativos à esquerda, ou seja, `8 / 4  / 2` é `(8 / 4) / 2`, e não `8 / (4 / 2)`.

Além disso, note que `+`/`-` podem ser operadores unários: `+3` (unário), `-4` (unário), `-(2+5)` (`-` unário, `+` binário), `-2+5` (`-` unário, `+` binário). Quando unários, `+`/`-` têm precedência sobre `*`, `/`, `//`, `%`.

Antes de executar, calcule mentalmente o valor de cada expressão abaixo e anote em comentários. Depois execute e confira.

````python
# Previsões:
# 2 ** 3 ** 2 ->
# 8 / 4 / 2 ->
# 100 // 7 % 3 ->
# 2 + 3 * 4 ** 2 // 5 % 3 ->
# (2 + 3) * (4 - 1) ** 2 // 5 ->
# -3 ** 2 ->
# (-3) ** 2 ->
# 1 -- 2 ->

print(2 ** 3 ** 2)
print(8 / 4 / 2)
print(100 // 7 % 3)
print(2 + 3 * 4 ** 2 // 5 % 3)
print((2 + 3) * (4 - 1) ** 2 // 5)
print(-3 ** 2)
print((-3) ** 2)
print(1 -- 2) # o que está acontecendo aqui?
````

Obs.: para muitos casos acima, é recomendado usar parênteses para não deixar dúvida!
